```python
import pandas as pd
import numpy as np

data = pd.read_csv("../data/Delivery_Logistics.csv")

data.head()
```

```python
data.isnull().sum()
```

```python
data.duplicated().sum()
```

```python
data = data.drop_duplicates()
```

```python
data["delivery_time_hours"] = pd.to_datetime(
    data["delivery_time_hours"]
).dt.nanosecond

data["expected_time_hours"] = pd.to_datetime(
    data["expected_time_hours"]
).dt.nanosecond
```

```python
data.dtypes
```

```python
data.describe()
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

numeric_columns = [
    "distance_km",
    "package_weight_kg",
    "delivery_time_hours",
    "expected_time_hours",
    "delivery_rating",
    "delivery_cost"
]

data[numeric_columns] = scaler.fit_transform(
    data[numeric_columns]
)
```
